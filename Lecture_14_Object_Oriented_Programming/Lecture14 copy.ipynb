A class is defined by specifying its:

1. **Constructor**. A function describing how to create an *instance* of a class.

2. **State**. Collection of objects a class instance contains. Each individual object is an "attribute". These are objects that an instance of a class contains, used to characterize that instance.

3. **Methods**. Functions connected to the class describing what actions an instance can perform.

We know that we can reduce $\frac{6}{8}$ to $\frac{3}{4}$. Wouldn't it be nice to just type `my_fraction.reduce()` and have it express itself in lowest terms? Let's write that method.

First, what is the algorithm for reducing a fraction (whether doing it on a computer or with pencil and paper)?

Once we know what to do, we will implement this method.

```python
print(33 / 4) # floating point division ("true division")
print(33 // 4) # integer division
print(33 % 4) # modulo division ("remainder")
```

```python
# Add reduce method

def gcd(m, n):
    # Euclid's algorithm
    # valid when n > 0
    while m % n != 0:
        m, n = n, m % n
    return n


class Fraction:
    """Class Fraction"""
    def __init__(self, input1, input2):
        self.num = input1 
        self.den = input2

    def show(self):
        print(f"{self.num}/{self.den}")

    def gcd(self, m, n):
        while m % n != 0:
            m, n = n, m % n
        return n

    def reduce(self):
        common = self.gcd(self.num, self.den)
        # common = gcd(self.num, self.den)

        self.num = self.num // common
        self.den = self.den // common
```

```python
my_fraction = Fraction(6, 8)
my_fraction.show()
my_fraction.reduce()
my_fraction.show()
```

```python
# add __str__ method

class Fraction:
    """Class Fraction"""
    def __init__(self, input1, input2):
        self.num = input1 
        self.den = input2

    def show(self):
        print(f"{self.num}/{self.den}")

    def gcd(self, m, n):
        while m % n != 0:
            m, n = n, m % n
        return n

    def reduce(self):
        common = gcd(self.num, self.den)

        self.num = self.num // common
        self.den = self.den // common

    def __str__(self):
        return f"{self.num}/{self.den}"
```

To define addition, we must include a method named `__add__`.

```python
# add the __add__ method


class Fraction:
    """Class Fraction"""
    def __init__(self, input1, input2):
        self.num = input1 
        self.den = input2

    def show(self):
        print(f"{self.num}/{self.den}")

    def gcd(self, m, n):
        while m % n != 0:
            m, n = n, m % n
        return n

    def reduce(self):
        common = gcd(self.num, self.den)

        self.num = self.num // common
        self.den = self.den // common

    def __str__(self):
        return f"{self.num}/{self.den}"

    def __repr__(self):
        return f"Fraction(num={self.num}, den={self.den})"

    def __add__(self, other_self):
        new_num = self.num * other_self.den + \
            self.den * other_self.num
        new_den = self.den * other_self.den

        common = gcd(new_num, new_den)
        return Fraction(new_num // common, new_den // common)
```

```python
f1 = Fraction(1, 4)
f2 = Fraction(1, 2)

print(f1.__add__(f2))
```

```python
f1 = Fraction(1, 4)
f2 = Fraction(1, 2)

print(f1 + f2)
```

While it is true that

$$ \frac{1}{4} + \frac{1}{2} = \frac{6}{8} $$

it would be much nicer if we divided both numerator and denominator by their greatest common divisor (which is 2 in this case).

Now, we could do something like:

```python
f1 = Fraction(1, 4)
f2 = Fraction(1, 2)

f3 = f1 + f2
f3.reduce()

print(f3)
```

<font size = "4">

But can we adjust the `__add__` method so that it automatically simplifies the fraction?

```python
# The previous __add__ already simplifies using gcd. This version makes the steps explicit.
def add_and_reduce(self, other):
    result = Fraction(self.num * other.den + other.num * self.den, self.den * other.den)
    result.reduce()
    return result
Fraction.__add__ = add_and_reduce
```

```python
f1 = Fraction(1, 4)
f2 = Fraction(1, 2)

print(f1 + f2)
```

Can override the `__eq__` method, so that it checks for deep equality instead of shallow equality

```python
# Add the __eq__ method

class Fraction:
    """Class Fraction"""
    def __init__(self, input1, input2):
        self.num = input1 
        self.den = input2

    def show(self):
        print(f"{self.num}/{self.den}")

    def gcd(self, m, n):
        while m % n != 0:
            m, n = n, m % n
        return n

    def reduce(self):
        common = gcd(self.num, self.den)

        self.num = self.num // common
        self.den = self.den // common

    def __str__(self):
        return f"{self.num}/{self.den}"

    def __repr__(self):
        return f"Fraction(num={self.num}, den={self.den})"

    def __add__(self, other_self):
        new_num = self.num * other_self.den + \
            self.den * other_self.num
        new_den = self.den * other_self.den

        common = gcd(new_num, new_den)
        return Fraction(new_num // common, new_den // common)

    def __eq__(self, other_self):
        left_num = self.num * other_self.den
        right_num = self.den * other_self.num
        return left_num == right_num
```

```python
f1 = Fraction(1, 2)
f2 = Fraction(2, 4)
print(f1 == f2) # check for deep equality
print()
print(id(f1) == id(f2)) # check for shallow equality
print(f1 is f2) # alternative
```

```python
print(f1 != f2)
```

**1.** Add a method called `__sub__` to `Fraction` so that `f1 - f2` works. The rule is

$$\frac{a}{b} - \frac{c}{d} = \frac{ad - cb}{bd}$$

Ensure the function expresses a simplified fraction, exactly as `__add__` does, and check `Fraction(3, 4) - Fraction(1, 4)`.

```python
# Complete exercise version: methods for Exercises 1–3 and 7–9 are grouped here.
class Fraction:
    def __init__(self, num, den):
        assert den != 0, "Denominator cannot be zero"
        if den < 0:
            num, den = -num, -den
        self.num = num
        self.den = den

    def gcd(self, m, n):
        while n != 0:
            m, n = n, m % n
        return abs(m)

    def reduce(self):
        common = self.gcd(self.num, self.den)
        self.num = self.num // common
        self.den = self.den // common

    def show(self):
        print(self)

    def __str__(self):
        return f"{self.num}/{self.den}"

    def __repr__(self):
        return f"Fraction(num={self.num}, den={self.den})"

    def __add__(self, other):
        result = Fraction(self.num*other.den + other.num*self.den, self.den*other.den)
        result.reduce()
        return result

    def __sub__(self, other):
        result = Fraction(self.num*other.den - other.num*self.den, self.den*other.den)
        result.reduce()
        return result

    def to_float(self):
        return self.num / self.den

    def to_percent(self, ndigits=3):
        return str(round(100*self.to_float(), ndigits)) + "%"

    def __mul__(self, other):
        result = Fraction(self.num*other.num, self.den*other.den)
        result.reduce()
        return result

    def __truediv__(self, other):
        assert other.num != 0, "Cannot divide by a zero fraction"
        result = Fraction(self.num*other.den, self.den*other.num)
        result.reduce()
        return result

    def __eq__(self, other):
        return self.num*other.den == other.num*self.den

    def __ne__(self, other):
        return not self == other

    def __gt__(self, other):
        return self.num*other.den > other.num*self.den

    def __ge__(self, other):
        return self.num*other.den >= other.num*self.den

    def __lt__(self, other):
        return self.num*other.den < other.num*self.den

    def __le__(self, other):
        return self.num*other.den <= other.num*self.den

print(Fraction(3, 4) - Fraction(1, 4))
```
