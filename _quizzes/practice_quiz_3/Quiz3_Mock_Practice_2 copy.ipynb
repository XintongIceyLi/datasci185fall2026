## A. 字典与 running maximum（4 分钟）

`club` 是名字到个人资料的嵌套字典。用 `for` 循环找出生年份最大的成员（最年轻的人）。

- 开始时使用下面给定的 `youngest_year` 和 `youngest_name`。
- 找到更大的年份时，同时更新年份和姓名。
- 不要使用 `max()` 或排序来代替循环。
- 输出 `The youngest member is ___, born in ___`。

检查结果应为 Hana、2009。

````python
club = {
    "Nora": {"year": 2001, "major": "Biology"},
    "Owen": {"year": 1997, "major": "Physics"},
    "Hana": {"year": 2009, "major": "English"},
    "Eli": {"year": 2004, "major": "Data Science"},
}
youngest_year = 0
youngest_name = "nobody"
````

````python
# Your code here
for people in club:
    if club[people]["year"]>youngest_year:
        youngest_year=club[people]["year"]
        youngest_name=people
print(f"The youngest member is {youngest_name}, born in {youngest_year}")
````

## B. 嵌套循环与动态分母（6 分钟）

随机选中 2–4 家商店。使用 `for` 循环，计算每一种商品在**选中商店**中的平均价格。

- 每种商品打印一行：`Average price of pen: ___ dollars`，四舍五入到两位小数。
- 商店数量会变，不能把分母写死为 3。
- 重新运行 setup 几次检查代码。

````python
import random
shop_prices = {
    "North": {"pen": 2.0, "notebook": 4.0, "folder": 1.5},
    "East": {"pen": 3.0, "notebook": 6.0, "folder": 2.5},
    "South": {"pen": 2.5, "notebook": 5.0, "folder": 2.0},
    "West": {"pen": 4.0, "notebook": 7.0, "folder": 3.0},
}
products = ["pen", "notebook", "folder"]
selected = random.sample(list(shop_prices.keys()), random.randint(2, 4))
selected_shops = {}
for shop in selected:
    selected_shops[shop] = shop_prices[shop]
print("Selected shops:", selected)
````

````python
# Your code here
for item in products:
    sum=0
    avg=0
    for shop in selected_shops:
        sum=sum+shop_prices[shop][item]
    avg=round(sum/len(selected_shops))
    print(f"Average price of {item}: {avg} dollars.")
````

## C. 将 while 循环封装成函数（5 分钟）

药物初始量为 800，每小时损失当前量的 `loss_rate`。当剩余量**小于或等于 400** 时停止。

把下面代码封装成函数 `hours_to_half(loss_rate)`，**返回**需要的小时数。每次调用都从 800 和 0 小时开始。

使用循环打印损失率 0.10、0.20、0.50 的结果，应依次得到 **7、4、1**。可以假设 `0 < loss_rate <= 1`。

````python
amount = 800
loss_rate = 0.10
hours = 0
while amount > 400:
    amount = amount * (1 - loss_rate)
    hours = hours + 1
print(hours)
````

````python
# Your code here
def hours_to_half(loss_rate):
    amount = 800
    loss_rate = 0.10
    hours = 0
    while amount > 400:
        amount = amount * (1 - loss_rate)
        hours = hours + 1
    return(hours)

print(hours_to_half(0.10))
````

## F. keyword-only 参数与坐标转换（5 分钟）

写函数 `point_distance(p1, p2, *, polar=False)`：

- 前两个参数是表示点的 tuple，可以按位置传入。
- `polar` 必须 keyword-only，默认 False；False 时按直角坐标处理。
- True 时两个 tuple 为 `(r, theta)`，theta 的单位是**弧度**；先转换，再计算距离。
- 转换公式：`x = r * np.cos(theta)`；`y = r * np.sin(theta)`。
- 可以调用下面已有的 `euclidean_distance`。

测试：`point_distance((1, 1), (4, 5))` 为 5；`point_distance((2, 0), (2, np.pi), polar=True)` 约为 4。

思考：为什么 `point_distance((2, 0), (2, np.pi), True)` 应报 TypeError？

````python
import numpy as np
def euclidean_distance(p1, p2):
    return np.sqrt((p1[0] - p2[0])**2 + (p1[1] - p2[1])**2)
````
