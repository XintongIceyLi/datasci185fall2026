## Quick Reference — Lecture 6

### 1. If / elif / else｜条件判断

```python
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
else:
    grade = "C or below"

print(grade)
```

一条 if/elif/else 链只执行第一个符合条件的分支。
每个条件行末有 `:`；分支内容要缩进。

### 2. Multiple if statements｜独立判断

```python
x = 8

if x > 5:
    print("Greater than 5")

if x < 10:
    print("Less than 10")
```

两个独立 if 都会检查；这里会打印两行。

### 3. And / or / not｜逻辑运算

```python
if x >= 5 and x <= 10:
    print("Between 5 and 10")

if x < 5 or x > 10:
    print("Outside the interval")

if not x == 8:
    print("Not 8")
```

区间也可写成：

```python
if 5 <= x <= 10:
    print("Between 5 and 10")
```

### 4. Nested if｜嵌套判断，不使用 and

```python
category = "student"
age = 20

if category == "student":
    if age < 18:
        price = 5
    else:
        price = 8
else:
    price = 12

print(price)
```

先判断类别，再判断该类别里的情况。
else 与它对应的 if 对齐。

### 5. Membership｜是否在列表或集合里

```python
month = "January"

if month in ["December", "January", "February"]:
    season = "Winter"
else:
    season = "Another season"
```

### 6. String comparison｜字符串比较

```python
name1 = "Amy"
name2 = "Ben"

if name1 < name2:
    print(f"{name1} comes before {name2}")
elif name1 > name2:
    print(f"{name2} comes before {name1}")
else:
    print("The names are the same")
```

按字符依次比较，区分大小写。

### 7. Common syntax｜常用写法

```python
x = 8          # 赋值
x == 8         # 判断相等
x != 8         # 判断不相等
x >= 8         # 大于或等于
x > 8          # 严格大于

print(f"The value is {x}")
```

**检查边界：** below → `<`；above → `>`；at least → `>=`；at most → `<=`。

````python
# create empty lists
good_movies = [] 
other_movies = []

cut_off = 8.0 # minimum rating for "good movie"

# Define variables for 3 movie titles and their corresponding IMDB rating
title_1 = "The Good, the Bad, the Weird"
imdb_rating_1 = 7.2 

title_2 = "Citizen Kane" 
imdb_rating_2 = 8.2 

title_3 = "Spider-Man: Across the Spider-Verse" 
imdb_rating_3 = 8.5 


# Divide movies between the two lists

if imdb_rating_1 >= cut_off:
    good_movies = good_movies + [title_1] # add movie to good_movies
else:
    other_movies = other_movies + [title_1] # add movie to other_movies

if imdb_rating_2 >= cut_off:
    good_movies = good_movies + [title_2] 
else:
    other_movies = other_movies + [title_2]

if imdb_rating_3 >= cut_off:
    good_movies = good_movies + [title_3]
else:
    other_movies = other_movies + [title_3]

print(good_movies)
print(other_movies)
````

````python
# create empty lists
good_movies = [] 
ok_movies = []
bad_movies = []

cut_off_1 = 8.0 # minimum rating for "good movie"
cut_off_2 = 6.0 # minimum rating for "ok movie"


# Define variables for 3 movie titles and their corresponding IMDB rating
title_1 = "The Good, the Bad, the Weird"
imdb_rating_1 = 7.2 

title_2 = "Citizen Kane" 
imdb_rating_2 = 8.2 

title_3 = "Spider-Man: Across the Spider-Verse" 
imdb_rating_3 = 8.5 

title_4 = "Melania"
imdb_rating_4 = 1.6

# Divide movies between the three lists
# print a warning if we encounter a bad movie

if imdb_rating_1 >= cut_off_1:
    good_movies = good_movies + [title_1] # add movie to good_movies
elif imdb_rating_1 >= cut_off_2:
    ok_movies = ok_movies + [title_1] # add movie to ok_movies
else:
    bad_movies = bad_movies + [title_1] # add movie to bad_movies
    print("Warning! Bad movie alert:", title_1) # ALSO: print warning message

if imdb_rating_2 >= cut_off_1:
    good_movies = good_movies + [title_2]  
elif imdb_rating_2 >= cut_off_2:
    ok_movies = ok_movies + [title_2]
else:
    bad_movies = bad_movies + [title_2]
    print("Warning! Bad movie alert:", title_2)


if imdb_rating_3 >= cut_off_1:
    good_movies = good_movies + [title_3] 
elif imdb_rating_3 >= cut_off_2:
    ok_movies = ok_movies + [title_3]
else:
    bad_movies = bad_movies + [title_3]
    print("Warning! Bad movie alert:", title_3)


if imdb_rating_4 >= cut_off_1:
    good_movies = good_movies + [title_4] 
elif imdb_rating_4 >= cut_off_2:
    ok_movies = ok_movies + [title_4]
else:
    bad_movies = bad_movies + [title_4]
    print("Warning! Bad movie alert:", title_4)


print(good_movies)
print(ok_movies)
print(bad_movies)
````

Birth season
<br>

- The cell below grabs a random person from `birthdays` 
- Write code that pulls out their month.
- Using the `in` keyword together with `if` / `elif` / `else`, decide which season that month belongs to and store the answer in a variable called `season`.
    - Winter: December, January, February
    - Spring: March, April, May

````python
person = random.choice(birthdays)
month = person[1]
if month in ["December", "January", "February"]:
    season = "Winter"
elif month in ["March", "April", "May"]:
    season = "Spring"
elif month in ["June", "July", "August"]:
    season = "Summer"
else:
    season = "Fall"
print(f"{person[0]} was born in {month}, which is in {season}.")
````
